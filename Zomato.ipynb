## Data Cleaning

```python
# reanme columns
data.rename(columns={"name":"restaurent",
                     "rate":"rating",
                     "cuisines":"food_type",
                     "listed_in(type)":"type",
                     "listed_in(city)":"city",
                     "approx_cost(for two people)":"cost"},inplace=True)
```

```python
replace = lambda x:x.replace('/5',"")
l = []
for val in map(replace,data['rating']):
    if val != 'NEW' and val != '-':
        var = float(val)
    l.append(var)
data['rating'] = l
```
